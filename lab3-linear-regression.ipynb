### Теория
Линейная регрессия моделирует зависимость между $X$ и $y$ как:
$$ y = Xw + \epsilon $$
Где $w$ — вектор весов (включая свободный член bias).

Аналитическое решение (OLS) находится по формуле:
$$ w = (X^T X)^{-1} X^T y $$
